```python
import ee
import geemap as g
from datetime import datetime, timedelta, date
import os
#import ollama
#import notebook
from IPython.core.display_functions import display
import ipywidgets as widget
from ipywidgets import HBox, VBox
from config import FIRMS_API_KEY

log = widget.display(
)
def ee_dateConverter(data): # Nasa FIRMS requires ISO format, using astablished dates already selected and used above and just convert to correct format
    output = date.fromtimestamp(data.getInfo()["value"]/1000)
    return output

def date_updater(start, end,past_start,past_end):
    global py_Date,ee_date,Today,Past,Past1,Past2
    try:
        py_Date = date.today() - timedelta(days=start)
        ee_date = ee.Date(py_Date.strftime("%Y-%m-%d"))  # epoch format
        Today = ee_date                     #   }-Short 10 day period for "current"  Data
        Past = ee_date.advance(end, 'day')  #   } Limited by NASA FIRMS of 10 days max

        Past1 = ee_date.advance(past_start, 'day') #   }-Short 10 day period to gather "past" Data
        Past2 = ee_date.advance(past_end, 'day') #   }
    except:
        print("date_Update error")
        py_Date = date.today()
        ee_date = ee.Date(py_Date.strftime("%Y-%m-%d"))
        Past = ee_date.advance(-10, 'day')
        Past1 = ee_date.advance(-20, 'day')
        Past2 = ee_date.advance(-30, 'day')

    output = [Today,Past,Past1,Past2]    
    return output

```
